VISUALIZACIÓN

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 1. Crear rangos de edad para que el gráfico sea legible (ej. de 10 en 10 años)
# Definimos los cortes de los 15 a los 95 años
bins = np.arange(15, 95, 10)
exitos_anteriores["rango_edad"] = pd.cut(exitos_anteriores["age"], bins=bins)

# 2. Contar cuántos éxitos hay por cada rango de edad
exitos_por_edad = (
    exitos_anteriores["rango_edad"].value_counts().sort_index().reset_index()
)
exitos_por_edad.columns = ["Rango de Edad", "Cantidad de Éxitos"]

# 3. Crear el gráfico de barras
plt.figure(figsize=(10, 6))
sns.barplot(
    data=exitos_por_edad,
    x="Rango de Edad",
    y="Cantidad de Éxitos",
    palette="viridis",
)

plt.title("Cantidad de Éxitos de la Campaña por Rango de Edad", fontsize=14)
plt.xlabel("Rango de Edad", fontsize=12)
plt.ylabel("Número de Éxitos", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

```python

sns.barplot(
    data=exitos_por_empleo,
    x="Cantidad de Éxitos",
    y="Empleo",  
    palette="viridis",
)


plt.title("Cantidad de Éxitos de la Campaña por Tipo de Empleo", fontsize=15, fontweight="bold")
plt.xlabel("Número de Éxitos", fontsize=12)
plt.ylabel("Tipo de Empleo", fontsize=12)

plt.tight_layout() 
plt.show()
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    exitos_estado_marital,
    labels=exitos_estado_marital.index,  
    autopct="%1.1f%%", 
    startangle=140,  
    colors=sns.color_palette(
        "viridis"
    ), 
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1,
        "antialiased": True,
    }
)


plt.title(
    "Distribución de Éxitos Anteriores por Estado Marital",
    fontsize=14,
    fontweight="bold",
)
plt.show()
```

```python
sns.barplot(
    data=exitos_estudios,
    x="Cantidad de Éxitos",
    y="Nivel de Estudios",
    palette="viridis",  
)


plt.title(
    "Éxitos en la Campaña Anterior según el Nivel de Estudios",
    fontsize=15,
    fontweight="bold",
)
plt.xlabel("Número de Éxitos Históricos", fontsize=12)
plt.ylabel("Nivel de Estudios", fontsize=12)


plt.tight_layout()
plt.show()
```
